```python
from typing import TypedDict
from langgraph.graph import START, END, StateGraph
```

```python
# 1.定义状态 State
class AgentState(TypedDict):
    message: str
```

```python
# 2.创建 hello_world 节点
def hello_world(state: AgentState) -> AgentState:
    """
    hello world node
    :param state: 修改前的状态
    :return: 修改后的状态
    """
    # 每个节点的业务逻辑
    state['message'] = 'hello world, ' + state['message']
    return state
```

```python
# 3.构建图（langgraph）
# builder = StateGraph(状态)
builder = StateGraph(AgentState)

# 3.1 往这个图中添加节点
# builder.add_node(节点的名字, 节点的函数)
builder.add_node("hello_world", hello_world)

# 3.2 给节点之间加上边
builder.add_edge(START, "hello_world")
builder.add_edge("hello_world", END)

# 编译图
graph = builder.compile()
```

```python
# 4.使用图
result = graph.invoke({
    "message": '阿苑'
})
print(result)
```
